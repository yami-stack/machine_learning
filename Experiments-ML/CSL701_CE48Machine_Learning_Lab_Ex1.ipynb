• Task 2: Load train.csv into a Pandas DataFrame using pd.read_csv().

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

uploaded = files.upload()

# Read the uploaded CSV file
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)
```

• Task 7: Generate summary statistics using df.describe(); record mean, minimum and
maximum values of Age and Fare.

```python
# Task 7: Generate summary statistics
print("Summary Statistics:")
print(df.describe())

# Record the mean, minimum, and maximum values of Age and Fare
print("\nAge Statistics:")
print("Mean Age :", df["Age"].mean())
print("Minimum Age :", df["Age"].min())
print("Maximum Age :", df["Age"].max())

print("\nFare Statistics:")
print("Mean Fare :", df["Fare"].mean())
print("Minimum Fare :", df["Fare"].min())
print("Maximum Fare :", df["Fare"].max())
```

**Phase 4: Data Manipulation & Cleaning**

• Task 9: Fill missing Age values with the median.

```python
# Task 9: Fill missing Age values with the median

# Calculate the median age
median_age = df["Age"].median()

# Fill missing values in the Age column
df["Age"] = df["Age"].fillna(median_age)

# Verify that there are no missing values in Age
print("Missing values in Age after filling:")
print(df["Age"].isnull().sum())

# Display the median age used
print("Median Age used to fill missing values:", median_age)
```

• Task 10: Filter surviving passengers (Survived == 1).

```python
# Task 10: Filter surviving passengers
survivors = df[df["Survived"] == 1]

# Display the first 10 surviving passengers
print("Surviving Passengers:")
print(survivors.head(10))

# Display the total number of survivors
print("\nTotal Number of Survivors:", survivors.shape[0])
```

• Task 11: Convert Fare column to a NumPy array.

```python
# Task 11: Convert Fare column to a NumPy array

fare_array = df["Fare"].to_numpy()

# Display the NumPy array
print("Fare NumPy Array:")
print(fare_array)

# Display the type and shape of the array
print("\nType:", type(fare_array))
print("Shape:", fare_array.shape)
```

• Task 12: Calculate maximum and average Fare using np.max() and np.mean().

```python
fare_array = df['Fare'].to_numpy()

# Calculate maximum and average Fare
max_fare = np.max(fare_array)
avg_fare = np.mean(fare_array)

print("Maximum Fare:", max_fare)
print("Average Fare:", avg_fare)
```

**Phase 5: Data Visualization**

• Task 13: Plot histogram of Age.

```python
# Plot histogram of Age
plt.figure(figsize=(8, 5))
plt.hist(df['Age'], bins=20, edgecolor='black')
plt.title('Histogram of Age')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.show()
```

• Task 14: Plot bar chart of Survival Count.

```python
survival_count = df['Survived'].value_counts()

# Plot bar chart
plt.figure(figsize=(6, 4))
plt.bar(survival_count.index, survival_count.values)
plt.title('Survival Count')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Number of Passengers')
plt.xticks([0, 1], ['Not Survived', 'Survived'])
plt.show()
```

• Task 15: Plot scatter graph of Age vs Fare.

```python
import matplotlib.pyplot as plt

# Plot scatter graph of Age vs Fare
plt.figure(figsize=(8, 5))
plt.scatter(df['Age'], df['Fare'])
plt.title('Age vs Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```
